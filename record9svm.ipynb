```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report

texts = [
    "Children play in the park",
    "Playing games is fun",
    "The boys play football",
    "The girls play with a ball",
    "Games keep children healthy",
    "Everyone enjoys playing",

    "Music is very enjoyable",
    "Music makes people happy",
    "Children love listening to music",
    "Music can be loud or soft",
    "Songs are a part of music",
    "Music is heard everywhere"
]
labels = [
    "Play", "Play", "Play", "Play", "Play", "Play",
    "Music", "Music", "Music", "Music", "Music", "Music"
]
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(texts)
X_train, X_test, y_train, y_test = train_test_split(
    X, labels, test_size=0.2, random_state=42
)
model = SVC(kernel="linear")
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
new_text = ["Football is a popular game"]
new_text_vector = vectorizer.transform(new_text)
prediction = model.predict(new_text_vector)
print("\nNew Text:", new_text[0])
print("Predicted Class:", prediction[0])
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report

texts=["India won the cricket match",
       "THe football team scored a goal",
       "The player hit a centuary",
       "THe team won the championship",
       "The match was exciting",
       "The bowler took five wickets",

       "Python is used for machine learning",
       "Artificial INtelligence is growing rapidely",
       "The computer uses a new processor",
       "ML models need data",
       "cloud computing provides online services",
       "The software application uses AI"]


labels=["sports","sports","sports","sports","sports","sports",
        "Technology","Technology","Technology","Technology","Technology","Technology"]

vectorizer=TfidfVectorizer()
X= vectorizer.fit_transform(texts)

X_train, X_test, y_train, y_test = train_test_split(
    X, labels, test_size=0.2, random_state=42
)

model=SVC(kernel='linear')
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:",accuracy)
print("\nClassification Report:")
print(classification_report(y_test,y_test))


new_text =["The cricket player scored a century"]
new_text_vector=vectorizer.transform(new_text)
prediction=model.predict(new_text_vector)
print("New Text:", new_text[0])
print("Predicted class:", prediction[0])
```
